## Galaxy clusters: the Universe's largest structures

- Collections of **hundreds to thousands of galaxies**, bound together by gravity
- Most of their mass (~85%) is **invisible dark matter**
- We study them at redshifts $z = 0.2$–$0.65$ (look-back time: 2–6 billion years)
- We use **background galaxies** (further away) as probes of the foreground cluster mass



## Reading the mass from distorted galaxy shapes

**Weak gravitational lensing**: mass bends light, stretching background galaxy shapes

The signal $\Delta\Sigma(R)$ — excess surface mass density at radius $R$ — encodes both:
- The total **mass** $M_{200}$
- The **concentration** $c_{200}$: how centrally peaked the dark matter is



> Higher concentration → steeper profile (yellow/high-$c$ curves above)


## The science question: does cluster history predict concentration?

The **magnitude gap** $\Delta m$ = brightness difference between the 1st and 2nd brightest cluster galaxy

- Large $\Delta m$ → the dominant galaxy "ate" its rivals → older, more relaxed cluster → **higher concentration?**
- We parametrise the slope as $\gamma$:
$$c \propto (\Delta m)^{\gamma}$$



> Colour = magnitude gap. Lines show the true relation for two $\Delta m$ values.


## Complication 1: clusters aren't spheres

Real dark matter halos are **triaxial ellipsoids** — elongated along one axis.
The 3D shape determines the projected lensing signal we observe.



> Each panel: a different random cluster shape and orientation.
> Floor heatmap: predicted lensing convergence $\kappa$ (how much light is bent).


## Simulated shape distribution

We draw cluster shapes from a realistic distribution consistent with N-body simulations:

- **Intermediate/major axis ratio** $b/a$: nearly uniform (any elongation possible)
- **Minor/major axis ratio** $c/a$: skewed toward flatter shapes
- **Line-of-sight angle** $\theta_{\rm los}$: isotropic ($\propto \sin\theta$)
- **Sky-plane ellipticity**: peaks around 0.3–0.4



## Complication 2: we don't always know where the centre is

Photometric cluster finders sometimes assign the wrong galaxy as the cluster centre.
**~36% of clusters** in our simulation are miscentred by up to ~1 Mpc.

This smears the lensing profile — ignoring it would bias the mass and concentration.



> Left: distribution of offset magnitudes. Right: spatial distribution on sky.


## Our simulated cluster sample

We simulate a **DES Year 3-like** survey: 5,000 clusters at $z = 0.2$–$0.65$



> The magnitude gap $\Delta m$ follows a realistic distribution. We inject a known $\gamma = 0.2$ signal.


## What we measure: one noisy profile per cluster

Each cluster gives a **weak lensing profile** $\Delta\Sigma(R)$ measured from ~few hundred background galaxies.
Individual profiles are very noisy — this is why traditional analyses *stack* hundreds of clusters.



> Triangles = miscentred clusters (lower signal, wider profile). Each panel is one cluster.


## Our approach: fit all clusters simultaneously

**Traditional (stacking):** average many profiles → fit one mean model → lose information about individual clusters

**Our approach (destacking):** fit all 5,000 profiles *at the same time* under a shared model

$$\ln\mathcal{L} = \sum_{i=1}^{N} \ln\mathcal{L}_i\bigl(\Delta\Sigma_i \mid M_i,\, c_i(\Delta m_i, \gamma)\bigr)$$

The shared parameter $\gamma$ is constrained by **all clusters simultaneously**,
using each cluster's own $\Delta m$ to predict its concentration.

We use the **nautilus** nested sampler to explore the posterior.


## Result: we can recover $\gamma$



> Blue histogram: posterior probability for $\gamma$.
> Red dashed: true value ($\gamma = 0.2$). Orange dashed: broad prior.
> **Recovered median $\hat{\gamma} = 0.38$, within ~1$\sigma$ of truth.**


## How stable is the estimate? Bootstrap resampling

We resample the 5,000 clusters (with replacement) 10 times and re-fit each resample.
Each colour is one bootstrap posterior — they should all peak near the truth.



> The spread of these posteriors quantifies our uncertainty on $\hat{\gamma}$.


## Calibration across many simulations

We run 10 independent simulations (each with fresh random clusters), each with 10 bootstrap draws.
**Do the error bars include the truth?**



> Each point: ensemble median minus truth. Error bar: $\sigma_{\rm boot}$.
> Pink band: overall scatter. **Calibration ratio = 0.91** (1.0 = perfectly calibrated).


## Ongoing work: scanning $\gamma$ from $-1$ to $+1$

**Can we recover $\gamma$ across its full plausible range?**

We are running a systematic scan:
- **11 values** of $\gamma_{\rm true}$: $-1.0, -0.8, -0.6, \ldots, +0.8, +1.0$
- **10,000 clusters** per simulation (2× more than previous tests)
- **50 bootstrap resamples** per gamma value
- Fully resumable — each checkpoint saved to disk

**Progress:** $\gamma = -1.0$ complete ($\hat{\gamma} = -0.55$, bias $+0.45$), running $\gamma = -0.8$

**Expected output:** a plot of $\gamma_{\rm true}$ vs $\hat{\gamma}$ revealing linearity, any
$\gamma$-dependent bias, and where the estimator loses sensitivity.

## Summary

| | |
|---|---|
| **Question** | Does cluster assembly history (magnitude gap) predict dark matter concentration? |
| **Method** | Simultaneous joint fit of 5,000–10,000 individual weak-lensing profiles |
| **Realism** | Triaxial cluster shapes + miscentering + realistic DES Y3 noise |
| **Result so far** | $\gamma$ recovered within ~1$\sigma$ of truth; bootstrap calibration ratio ~0.9 |
| **Now running** | Full $\gamma$ scan from $-1$ to $+1$ to map estimator performance |

### Key advantages over stacking
- Retains per-cluster information (shape, Δm, redshift)
- Direct inference on the scaling relation slope γ
- Error bars from bootstrap rather than jackknife
